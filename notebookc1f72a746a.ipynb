```python
# ============================================================
# ACTIVITY 2: Compare Sampling Temperatures
# ============================================================
from transformers import pipeline

# Load pretrained GPT-2
generator = pipeline("text-generation", model="gpt2")

# Prompt
prompt = "The spaceship's engines failed just as"

# Generate at 3 temperatures
for temp in [0.2, 0.7, 1.3]:
    result = generator(
        prompt,
        max_new_tokens=40,
        do_sample=True,
        temperature=temp,
        truncation=True
    )
    print(f"Temperature {temp}:")
    print(result[0]["generated_text"])
    print("-" * 60)
```

```python
# ============================================================
# OPTIONAL EXTENSION: Your own prompts (text + image)
# ============================================================
from transformers import pipeline
from diffusers import StableDiffusionPipeline
import torch, matplotlib.pyplot as plt

# Load models
generator = pipeline("text-generation", model="gpt2")
pipe = StableDiffusionPipeline.from_pretrained(
    "stabilityai/sd-turbo", torch_dtype=torch.float16
).to("cuda")

# Your prompts
my_text_prompt  = "In a world where memories can be traded like currency,"
my_image_prompt = "An astronaut riding a horse on the surface of the sun, cinematic lighting"

# Text output
print("=== TEXT ===")
print(generator(my_text_prompt, max_new_tokens=60, truncation=True)[0]["generated_text"])

# Image output
print("\n=== IMAGE ===")
img = pipe(prompt=my_image_prompt, num_inference_steps=4, guidance_scale=0.0).images[0]
plt.imshow(img); plt.axis("off"); plt.show()
```
