```python
# ============================================================
# LAB TASK 1: Generate and Evaluate Story Completions
# Dataset: rathachat/writing-prompts
# Path: /kaggle/input/datasets/ratthachat/writing-prompts
# Stages: Load & Inspect → Load Model → Generate → Evaluate
# ============================================================
import os
from transformers import pipeline

# ---------- STAGE 1: LOAD & INSPECT ----------
DATASET_DIR = "/kaggle/input/datasets/ratthachat/writing-prompts"

# 1a. Show what's actually inside the dataset folder
print("=== CONTENTS OF DATASET FOLDER ===")
for root, dirs, files in os.walk(DATASET_DIR):
    print("DIR :", root)
    for f in files:
        print("   FILE:", f)
print()

# 1b. Auto-detect train.wp_source anywhere under the dataset folder
prompts_path = None
for root, dirs, files in os.walk(DATASET_DIR):
    for f in files:
        if f == "train.wp_source":
            prompts_path = os.path.join(root, f)
            break
    if prompts_path:
        break

# 1c. Fallback: if not found by exact name, try any .wp_source file
if prompts_path is None:
    for root, dirs, files in os.walk(DATASET_DIR):
        for f in files:
            if f.endswith(".wp_source"):
                prompts_path = os.path.join(root, f)
                break
        if prompts_path:
            break

if prompts_path is None:
    raise FileNotFoundError(
        "train.wp_source not found. Check the printed folder contents above "
        "and update the filename accordingly."
    )

print("Found prompts file at:", prompts_path, "\n")

with open(prompts_path, "r", encoding="utf-8") as f:
    all_prompts = f.readlines()

sample_prompts = [p.strip() for p in all_prompts[:5]]

print("=== SAMPLE PROMPTS ===")
for i, p in enumerate(sample_prompts):
    print(f"{i+1}. {p}")

# ---------- STAGE 2: LOAD A PRETRAINED MODEL ----------
print("\n=== LOADING GPT-2 ===")
generator = pipeline("text-generation", model="gpt2")

# ---------- STAGE 3: GENERATE ----------
print("\n=== GENERATING COMPLETIONS ===")
completions = []
for p in sample_prompts:
    result = generator(
        p,
        max_new_tokens=60,
        num_return_sequences=1,
        truncation=True
    )
    completions.append(result[0]["generated_text"])

for i, c in enumerate(completions):
    print(f"\n--- Completion {i+1} ---")
    print(c)

# ---------- STAGE 4: EVALUATE (manual) ----------
print("\n=== NOW SCORE EACH COMPLETION BY HAND USING THE RUBRIC BELOW ===")
print("""
Rubric: Coherence (1-5) | Relevance (1-5) | Factual Plausibility (1-5) | Notes
No 1: __ | __ | __ | ______________________
No 2: __ | __ | __ | ______________________
No 3: __ | __ | __ | ______________________
No 4: __ | __ | __ | ______________________
No 5: __ | __ | __ | ______________________
""")
```
