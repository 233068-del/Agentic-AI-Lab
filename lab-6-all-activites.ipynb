```python
import random

random.seed(7)


def print_banner(title):
    width = 60
    print("=" * width)
    print(title.center(width))
    print("=" * width)


class Environment:
    def __init__(self, size):
        self.size = size
        self.dirt = [random.choice([True, False]) for _ in range(size)]
        self.agent_position = 0

    def get_percept(self):
        return {
            "position": self.agent_position,
            "is_dirty": self.dirt[self.agent_position],
        }

    def apply_action(self, action):
        if action == "Suck":
            self.dirt[self.agent_position] = False
        elif action == "MoveRight" and self.agent_position < self.size - 1:
            self.agent_position += 1
        elif action == "MoveLeft" and self.agent_position > 0:
            self.agent_position -= 1

    def is_clean(self):
        return not any(self.dirt)

    def render(self):
        row = ""
        for i in range(self.size):
            cell = "D" if self.dirt[i] else "."
            if i == self.agent_position:
                cell = "[" + cell + "]"
            row += cell
        return row


if __name__ == "__main__":
    print_banner("ACTIVITY 1: BUILD THE ENVIRONMENT")
    env = Environment(size=5)

    print("\n--- INITIAL STATE ---")
    print(f"  Grid size        : {env.size}")
    print(f"  Agent position   : {env.agent_position}")
    print(f"  Dirt layout      : {env.dirt}")
    print(f"  Rendered world   : {env.render()}")

    print("\n--- PERCEPT ---")
    percept = env.get_percept()
    for key, value in percept.items():
        print(f"  {key:<10}: {value}")

    print("\n--- ENVIRONMENT STATUS ---")
    print(f"  Is the world clean? {env.is_clean()}")
```

```python
import random

random.seed(7)


def print_banner(title):
    width = 60
    print("=" * width)
    print(title.center(width))
    print("=" * width)


class Environment:
    def __init__(self, size):
        self.size = size
        self.dirt = [random.choice([True, False]) for _ in range(size)]
        self.agent_position = 0

    def get_percept(self):
        return {
            "position": self.agent_position,
            "is_dirty": self.dirt[self.agent_position],
        }

    def apply_action(self, action):
        if action == "Suck":
            self.dirt[self.agent_position] = False
        elif action == "MoveRight" and self.agent_position < self.size - 1:
            self.agent_position += 1
        elif action == "MoveLeft" and self.agent_position > 0:
            self.agent_position -= 1

    def is_clean(self):
        return not any(self.dirt)

    def render(self):
        row = ""
        for i in range(self.size):
            cell = "D" if self.dirt[i] else "."
            if i == self.agent_position:
                cell = "[" + cell + "]"
            row += cell
        return row


def simple_reflex_agent(percept, size):
    if percept["is_dirty"]:
        return "Suck"
    elif percept["position"] < size - 1:
        return "MoveRight"
    else:
        return "NoOp"


if __name__ == "__main__":
    print_banner("ACTIVITY 3: RUN AND LOG THE AGENT LOOP")

    env = Environment(size=5)
    print(f"\nStart world: {env.render()}")
    print("\n" + "-" * 60)
    print(f"{'Step':<6}{'Position':<10}{'Dirty?':<8}{'Action':<12}{'World'}")
    print("-" * 60)

    for step in range(10):
        percept = env.get_percept()
        action = simple_reflex_agent(percept, env.size)
        env.apply_action(action)

        print(f"{step+1:<6}{str(percept['position']):<10}"
              f"{str(percept['is_dirty']):<8}{action:<12}{env.render()}")

        if env.is_clean() and percept["position"] == env.size - 1:
            print("-" * 60)
            print(f"[DONE] All clean after {step+1} steps. Agent stops.")
            break
    else:
        print("-" * 60)
        print("[STOP] Reached max steps without fully cleaning.")
```
